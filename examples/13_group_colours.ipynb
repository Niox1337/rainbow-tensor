# Follow reduction groups by colour

A reduction combines several source values into one output value. Their matching backgrounds help you follow that connection. This example has 12 output groups and moves the focus between two of them.

```python
import numpy as np

import rainbow_tensor as rt

rt.set_language("en")
x = np.arange(24).reshape(2, 12)
x
```

## One column becomes one output

`axis=0` adds the two rows, leaving one output for each of the 12 columns. The two source cells in a column and their result use the same background.

`focus=(0,)` selects the first output. Its source values are 0 and 12, so the result is 12. The focused group uses a green background and bold values.

```python
first = rt.sum(x, axis=0, focus=(0,), theme="light")
first
```

## Move the focus without moving the groups

Coordinates start at zero, so `(6,)` selects the seventh output. The focus now marks 6 and 18 in the source and 24 in the result. The other groups keep their colours. The first group returns to its ordinary colour.

```python
middle = rt.sum(x, axis=0, focus=(6,), theme="light")
middle
```

```python
print("Source values for output (6,):", x[:, 6])
print("Their sum:", x[:, 6].sum())
print("Focused output coordinate:", middle.trace.output_coord)
```

## Try a fixed dark theme

The previous figures use a fixed light theme. This one uses a fixed dark theme. The source and result still share group colours, with shades chosen for the darker background.

```python
rt.sum(x, axis=0, focus=(6,), theme="dark")
```

## Let the viewer choose

`theme="auto"` follows the browser or operating system colour preference. Try switching that preference while this figure is open. The palette changes while the group matching and focused coordinate stay the same. AUTO is also the default when you have not set another theme.

```python
rt.sum(x, axis=0, focus=(6,), theme="auto")
```

## Keep dimensions while reducing two axes

Now the source has shape `(2, 12, 3)`. Reducing axes 0 and 2 combines six source values for each position on axis 1.

`keepdims=True` leaves the reduced axes at length 1, giving a result shape of `(1, 12, 1)`. The focus must include those axes, so the seventh group is `(0, 6, 0)`. Follow its six highlighted source cells to the single highlighted result.

```python
cube = np.arange(72).reshape(2, 12, 3)
kept = rt.sum(
    cube, axis=(0, 2), keepdims=True, focus=(0, 6, 0), theme="auto"
)
kept
```

```python
print("Result shape:", kept.result_shape)
print("Source values for output (0, 6, 0):")
print(cube[:, 6, :])
print("Their sum:", cube[:, 6, :].sum())
```

## Use coordinates when colours get close

The group palette grows beyond a short repeating list, but a screen cannot provide an unlimited number of easily distinguishable colours. With many groups, some shades will look similar.

Use a cell's coordinate and `focus` to identify the exact group. Hover over a cell to read its coordinate, then pass the corresponding output coordinate as `focus`.
