# Theme and language preferences

A figure can follow the viewer while its explanations follow the Python kernel language. These settings are independent. Explicit settings make a shared lesson reproducible.

```python
import numpy as np
from IPython.display import display

import rainbow_tensor as rt

x = np.arange(24).reshape(2, 3, 4)
print("Requested language:", rt.get_language())
print("Resolved catalog:", rt.get_resolved_language())
print("Available catalogs:", rt.available_languages())
rt.shape(x)
```

## Compare viewer preferences and fixed colours

The automatic figure responds to the browser or operating system colour preference without reading the tensor again. Fixed themes stay fixed. Save an automatic figure to keep this behaviour in an SVG viewer that supports CSS.

```python
rt.set_default_theme("auto")
display(rt.shape(x))
display(rt.shape(x, theme="light"))
display(rt.shape(x, theme="dark"))
```

## Select a language for the next figure

The regional tag `zh-CN` falls back to the shipped `zh` catalog. Coordinates and operation names stay unchanged. Existing figures keep the language in which they were created.

```python
rt.set_language("zh-CN")
assert rt.get_language() == "zh-CN"
assert rt.get_resolved_language() == "zh"
display(rt.sum(x, axis=(0, 2), keepdims=True, focus=(0, 1, 0)))
display(rt.shape((2, 0, 3)))
```

## Add translations without changing Python

Create a UTF-8 JSON file named for its language, for example `fr.json`. Copy message keys from `src/rainbow_tensor/locales/en.json` and translate their values. Keep placeholders unchanged. Missing entries use English. Load the file with `rt.load_translations("translations/fr.json")`, then select it with `rt.set_language("fr")`. The [translation guide](../docs/guide/translations.md) includes a complete example and validation rules.

```python
rt.set_language("en")
rt.set_default_theme("light")
rt.mean(x, axis=1, focus=(1, 2))
```

## Resume automatic settings

Automatic language selection reads the kernel environment and system locale. In a remote notebook, those settings belong to the server. Use a manual language override when the server language differs from your own.

```python
rt.set_language("auto")
rt.set_default_theme("auto")
print("Language:", rt.get_resolved_language())
print("Theme:", rt.get_default_theme().name)
```
